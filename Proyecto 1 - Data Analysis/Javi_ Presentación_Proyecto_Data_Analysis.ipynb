# CONCLUSIONES Y SIGUIENTES PASOS

· La tendencia de la participacion ciudadana es creciente

· La covid impulsó el registro de peticiones mediante WEB, sin embargo,  existe una tendencia a los medios tradicionales

· Para trámites administrativos se usa los canales WEB y para quejas que tienen que ver con el espacio el canal más usado es el teléfono

· En los dos años analizados Ciutat vella queda a la cabeza en nº de registros y Sarrià-St Gervasi quedan a la cola

· La limpieza es una preocupacion para los barceloneses con especial foco en el distrito de Ciutat Vella
