# OPERACIONES CON DF

1) Selecciona una columna:
- df["colunma_1"]: Si se hace con un par de corchetes devuelve la columna en forma de lista
- df[["colunma_1]]: Si se hace con dos corchetes devuelve un DF
- df[['columna_1' , 'columna_2']]: Se pueden selecciona más de una columna y se genera DF
   
2) Crear colmna nueva dentro de un mismo DF:
- df["nueva_columna"]=df['columna_x'].aplicamos_método : Nótese que se puede aplicar el métrodo a la columna existente a la derecha del igual
- df["nueva_columna"]=df['columna_x'] * df['columna_y'] : Se pueden hacer operaciones entre columnas (+, -, *, **, /, //)

3) Filtrados:
- df.columna_1.método: DataFrame.nombre_columna.método y devulve un array
- df[df['columna_1]=='algo']: filtra el DataFrame cumpliendo la condición que sea igual a algo y pueder ser (!= , < , >)--> df[condición].método
- df[df.columna_1 == 'algo']: devuelve lo mismo que l punto anterior
- df[(df['columna_1]=='algo') & (df['columna_2]!='algo')].método --> df[(cond_1) (operador) (cond_2) (operador) ....].método

## RENOMBRANDO DE COLUMNAS


1) MÉTODO PARA RENOMBRAR TODAS LAS COLUMNAS
- Guardamos las columnas en una nueva lista utilizando el método .columns y se transforma en list porque el método devuelve un array de nunpy

    columnas=list(df.columns)

- Se genera una nueva variable -->

     nuevas_columnas={columna: columna.replace(" ", "_").lower() for columna in columnas}

      El resultado es un diccionatio en el que se ha sustituido el espacio " " de los títulos por "_". Siendo las claves el valor antiguo y el modificado el nuevo valor.

- Se utiliza el método rename para renombra las columnas:
  
      df.rename(columns=nuevas_columnas, inplace=True) --> Se utiliza inplace=True para sobreescribir el resultado


2. MÉTODO PARA RENOMBRAR UNA O ALGUNA
   
- Generamos un dicionario con los nombre de las columnas que que queremos cambiar con la clave siendo el valor antiguo y la información el nuevo.
  
      para_unnamed = {"titulo_antiguo": "titulo_nuevo"}
  
      df.rename(columns=para_unnamed, inplace=True) --> Se aplica inplace=True para sobreescribir
  

3. EJEMPLO PASAR TODAS LAS COLUMNAS A MAYÚSCULAS / MINÚSCULAS

    mayus_col={columna: columna.upper() for columna in list(df.columns)} --> .lower() para poner minúsculas
   
    df.rename(columns=mayus_col , inplace=True)

4. EJEMPLO USANDO EL MÉTODO COLUMNS

- Creamos el dicionario con los las etiquetas de los títulos que queremos. P.e:
  
        lista_minus={columna: columna.lower() for columna in list(df.columns)}

        df.columns=lista_minus

## SELECCIÓN DE FILAS Y COLUMNAS

1. DF.ILOC[  ,  ] (i es de index y loc de location):


   - df.iloc[ filas , columnas ], donde  filas y  columnas son la posición de las filas y columnas que se desean seleccionar en el orden que aparecen en el objeto.
     
   - Si se pasa un solo argumento interpreta que se está omitienedoen segundo argumento (columnas) y devuelve la fila entera (todas las columnas de la fila)
  
   - Se puede elegir rangos df[a:b , c:d]: de la misma manera que antes si se elige solamente un rango devuelve todas las columnas en ese ranog de filas
  
   - Si queremosseleccionar unas columnas determinadas se hace d ela siguiente manera df.[a:b, [c,k]] y solamente se mostrarán las columna c y k
  


2. DF.LOC[  ,  ]: Se puede utilizar para selcciona filas o columnas en base a etiquetas o en base a una condición:


    - df.loc['valor_ind'] --> filtra filas con índice=valor_ind
      
    - df.loc[(df['columna_1]=='algo') & (df['columna_2]!='algo')] --> df[(cond_1) (operador) (cond_2) (operador) ....] --> Devuelve todas las filas que cumplan las condiciones
  
    - df.loc[(df['columna_1]=='algo') & (df['columna_2]!='algo'), [columna_x , columna_y, columna_z]] --> PErmite aplicar la condicion y mostrar únicamente determinadas columnas

## AGRUPAR

1. df.groupby(df[column_x]).mean(): Agrupa por los .unique valores de column_x y calcula la media para todas las columnas según la categoria
   
2. df.groupby(df[column_x, column_y]).mean(): Agrupa siguiendoselos el mismo criterio siendonosla column_x primer nivel(más a la izquierda) y column_y segundo nivel (a continuacion)
   
3. df.groupby("column_x").agg({"columna_otra_1": "mean", "columna_otra_2": "sum"}): Agrupa en base a las categoria de la column_x y en columna_otra_1 calcula la media y en columna_otra_2 calcula la suma según categórias en column_x

4. df.groupby(["column_x", "column_y").agg({"columna_otra_1": "mean", "columna_otra_2": "sum"}): Lo mismo que punto 3 pero poniendo un segundo nivel de categorias

Ejemplo: titanic["age"] = titanic.groupby(["sex", "class"])["age"].apply(lambda x: x.fillna(x.mean()))

Rellena los valores vacíos con la media de edad según la categorízación de sexo y clase. Recuerda que ["age"] es para seleccionar sólamente la columna ["age"]
