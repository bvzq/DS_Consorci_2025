```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

import time
import pandas as pd
from selenium import webdriver
#from selenium.webdriver.chrome.service import Service
from selenium.webdriver.safari.service import Service
from selenium.webdriver.common.by import By
#from selenium.webdriver.chrome.options import Options
from selenium.webdriver.safari.options import Options
```

```python
# Iniciar el navegador
#driver = webdriver.Chrome()
driver = webdriver.Safari()
# URL principal de Carrefour Supermercado
url = "https://www.thomann.de/es/cat.html"
# url = "https://www.thomann.de/es/index.html"
driver.get(url)
time.sleep(5)

links_elements = driver.find_elements(By.TAG_NAME, "a")

links = set()

# Recorrer los elementos y obtener el texto y el href
for link in links_elements:
    try:
        url = link.get_attribute("href")
        if url:  # Solo guardar si el enlace no está vacío
            links.add(url)
    except:
        continue

# Cerrar el navegador
driver.quit()
```

```python
palabras_clave = ['accesorio','altavoz','ampli','micro','mesas','material','guitarr','libros']
# clases_articulos = ['amplificad']
articulos = []
for clases in palabras_clave:
    for link in sorted(links):  # Ordenar alfabéticamente
        if clases in link:
            articulos.append(link)
articulos = set(articulos)
print(len(articulos))
```

```python
from bs4 import BeautifulSoup
import pandas as pd

# Extract all product containers
products = soup.find_all("a", class_="fx-product-box")

# List to store extracted data
product_data = []

# Loop through each product and extract information
for product in products:
    # Extract product name from the <img> alt attribute
    name = product.find("img", class_="fx-image")
    name = name["alt"] if name else "N/A"

    # Extract manufacturer (brand)
    manufacturer = product.find("span", class_="description__manufacturer")
    manufacturer = manufacturer.get_text(strip=True) if manufacturer else "N/A"

    # Extract product price
    price = product.find("span", class_="fx-typography-price-primary")
    price = price.get_text(strip=True) if price else "N/A"

    # Extract rating count
    rating_count = product.find("div", class_="fx-rating-stars__description")
    rating_count = rating_count.get_text(strip=True) if rating_count else "N/A"

    # Extract product URL
    product_url = f"https://www.thomann.de/{product['href']}" if product.has_attr("href") else "N/A"

    # Append extracted data to the list
    product_data.append({
        "Product Name": name,
        "Manufacturer": manufacturer,
        "Price": price,
        "Rating Count": rating_count,
        "Product URL": product_url
    })

# Convert list to DataFrame
df_products = pd.DataFrame(product_data)

# ✅ Display DataFrame in Jupyter Notebook or Terminal
print(df_products)  # For console output

# ✅ If using Jupyter Notebook, display it nicely
try:
    from IPython.display import display
    display(df_products)
except ImportError:
    pass  # If IPython is not available, fallback to print
```
