##  Campo calculado: identificando la fuente de la fila y las tablas

[Algunos documentos](https://help.tableau.com/current/pro/desktop/en-us/functions_operators.htm)

- Llámalo "sexo"

Una opción: `IF ([Nombre de tabla] = "hombres.csv") Then "hombres" ELSE "mujeres" END`

- Escribe esta fórmula: `IIF ([Nombre de tabla] = "hombres.csv", "hombres", "mujeres")`

##  Tabla pivote: de formato ancho a largo

- Asegúrate de cambiar `Población` a Medida

- Ahora arrastra `Provincias` al menú de Marcas una vez a Color y otra vez a Etiqueta
